```python
import numpy as np
h = 6.62607015e-34 
c = 2.99792458e8  
wavelengths_nm = np.array([405, 488, 532, 633, 650])  
wavelengths_m = wavelengths_nm*1e-9 
energies_J = h*c/wavelengths_m
powers_mW = np.array([12.5, 15.2, 18.7, 14.1, 11.8])
powers_W = powers_mW*1e-3
N = powers_W/energies_J
print(wavelengths_m)
print("photon energies (J):", energies_J)
print("powers (W):", powers_W)
print("photon rates (photons/s):", N)
```

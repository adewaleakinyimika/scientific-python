```python
h = 6.62607015e-34
c = 2.99792458e8
def photon_energy(wavelength_nm):
    wavelength_m = wavelength_nm*1e-9
    E = h*c/wavelength_m
    return E
```

```python
try:
    wavelength_nm = float(input("enter the wavelength in nm: "))
    Energy = photon_energy(wavelength_nm)
    print(f"Photon energy = {Energy:.2e} J")
except ValueError:
    print("invalid wavelength. Enter a numerial value")
```

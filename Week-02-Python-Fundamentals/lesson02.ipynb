```python
import math
h = 6.62607015e-34
c = 2.99792458e8

def photon_energy(wavelength_nm):
    wavelength_m = wavelength_nm*1e-9
    E = h*c/wavelength_m
    return E

def critical_angle(n1,n2):
    if n1>n2:
        theta = math.asin(n2/n1)
        theta_d = math.degrees(theta)
        return theta_d
    else:
        return None

def image_properties(f,d_o):
    if f==d_o:
        return None,None
    else:
        d_i = 1/(1/f - 1/d_o)
        M = -d_i/d_o
        return d_i, M

def optical_resolution(wavelength_nm,NA):
    if NA <= 0:
        return None
    else:
        x_m = wavelength_nm*1e-9
        d_m = 0.61*x_m/NA
        return d_m

def rayleigh_range(wavelength_nm, waist_mm):
    wavelength_m = wavelength_nm*1e-9
    waist_m = waist_mm*1e-3
    z_r = math.pi*waist_m**2/wavelength_m
    return z_r



while True:
    print("Optics Calculator: what do you need? select the number from the list below:")
    print("1. Photon energy")
    print("2. Critical angle")
    print("3. Thin Lens")
    print("4. Optical resolution")
    print("5. Rayleigh range")
    print("0. Exit")

    choice = int(input("enter the number: "))
    if choice == 0:
        print("Calculator closed.")
        break
    elif choice == 1:
        wavelength_nm = float(input("enter wavelength in nm: "))
        energy = photon_energy(wavelength_nm)
        print(f"photon_energy = {energy:.3e} J")
    elif choice == 2:
        n1 = float(input("enter n1: "))
        n2 = float(input("enter n2: "))
        theta_c = critical_angle(n1,n2)
        if theta_c is not None:
            print(f"The critical angle = {theta_c:.2f}")
        else:
            print("Total internal reflection is not possible")
    elif choice == 3:
        f = float(input("enter the focal length: "))
        d_o = float(input("enter the focal object distance: "))
        image_d, mag = image_properties(f,d_o)
        if image_d is not None:
            print(f"Image distance = {image_d:.2f} cm, Magnification = {mag:.2f}")
        else:
            print("Image is formed at infinity")
    elif choice == 4: 
        wavelength_nm = float(input("enter wavelength in nm: "))
        NA = float(input("enter numerical aperture: "))
        res = optical_resolution(wavelength_nm, NA)
        if res is None:
            print("Numerical aperture must be greater than zero")
        else:
            res_nm = res*1e9
            print(f"resolution in meter = {res:.2e} m, resolution in nm = {res_nm:.2e} nm")

    elif choice == 5:
        wavelength_nm = float(input("enter wavelength in nm: "))
        waist_mm = float(input("enter beam-waist radius in mm: "))
        rayleigh_m = rayleigh_range(wavelength_nm, waist_mm)
        print(f"Rayleigh range = {R_range:.3f} m")
    else:
        print("Invalid selection. Please enter a number from 0 to 5.")
```

```python
def beam_radius(lamb_nm, wist_mm, z):
    wist_m = wist_mm*1e-3
    z_r = rayleigh_range(lamb_nm,wist_mm)
    w_z = wist_m*math.sqrt(1 + (z/z_r)**2)
    return w_z
beam_rad = beam_radius(532, 0.5, 1.476)
w_z_mm = beam_rad*1e3
print(f"Beam radius = {w_z_mm:.3f} mm")
```

```python
wavelentgh_nm = float(input("enter wavelength in nm: "))
power_mW = float(input("enter Laser power in mW: "))
diameter_mm = float(input("enter Beam diameter in mm: "))
wavelength_m = wavelentgh_nm*1e-9
power_W = Power_mW*1e-3
Diameter_m = Diameter_mm*1e-3
c = 2.99792458e8 # m/s
h = 6.62607015e-34 # Js
pi = 3.141592653589793
f = c/wavelength_m
E = h*f
A = pi*(Diameter_m/2)**2
I = power_W/A
print(wavelentgh_nm, type(wavelentgh_nm))
print(Power_mW, type(Power_mW))
print(Diameter_mm, type(Diameter_mm))
print(wavelentgh_nm > 500)
print(Power_mW == 10)
print(Diameter_mm != 1)
print(f"frquency = {f:.2e} Hz", type(f))
print(f"photon energy = {E:.2e} J", type(E))
print (f"Beam area = {A:.2e} m^2", type(A))
print(f"Intensity = {I: 2e} W/m^2", type(I))
```
